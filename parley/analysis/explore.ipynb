# Explore a Parley match

Export a match first: `uv run parley export <match_id> --out analysis/<match_id>.jsonl` (match ids: `uv run parley matches`).
Each row below is one role in one episode.

```python
import glob
import matplotlib.pyplot as plt
from parley.analysis import compare, load_episodes

PATHS = sorted(glob.glob('*.jsonl'))  # every export in this folder
df = load_episodes(*PATHS)
df.head()
```

## Payoff by model, with 95% bootstrap confidence intervals

```python
table = compare(df, 'model')
table
```

```python
fig, ax = plt.subplots(figsize=(6, 3))
ax.errorbar(table['model'], table['mean'],
            yerr=[table['mean'] - table['ci_low'], table['ci_high'] - table['mean']], fmt='o', capsize=4)
ax.set_ylabel('mean payoff'); ax.set_title('Payoff by model (95% CI)')
plt.xticks(rotation=20, ha='right'); plt.tight_layout()
```

## Payoff by model and role

Roles are swapped between runs, so each model is seen in each role.

```python
compare(df, ['model', 'role'])
```

## Outcomes and metrics

```python
df.groupby('model')[['deal']].mean().rename(columns={'deal': 'deal_rate'})
```

```python
metric_cols = [c for c in df.columns if c.startswith('m.')]
df.groupby('model')[metric_cols].mean().T
```
